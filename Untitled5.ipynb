```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler

pd.set_option('display.max_columns', None)
sns.set(style="whitegrid")
%matplotlib inline
```

```python
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/catholic.csv"
df = pd.read_csv(url)

print("Shape of dataset:", df.shape)
df.head()
```

```python
df.info()
```

```python
df.describe().T
```

```python
summary = pd.DataFrame({
    "dtype": df.dtypes,
    "n_unique": df.nunique(),
    "n_missing": df.isnull().sum()
})
summary
```

```python
missing = df.isnull().sum()
missing_percent = (missing / len(df)) * 100
missing_df = pd.DataFrame({"Missing Count": missing, "Missing %": missing_percent})
missing_df = missing_df[missing_df["Missing Count"] > 0].sort_values("Missing %", ascending=False)
missing_df
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, cmap="viridis")
plt.title("Missing Values Heatmap")
plt.show()
```

```python
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df.select_dtypes(exclude=[np.number]).columns.tolist()

for col in num_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

for col in cat_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Total missing values after imputation:", df.isnull().sum().sum())
```

```python
if "rownames" in df.columns:
    df.drop(columns=["rownames"], inplace=True)

# Check and remove duplicate rows
print("Duplicate rows before cleaning:", df.duplicated().sum())
df.drop_duplicates(inplace=True)
print("Duplicate rows after cleaning:", df.duplicated().sum())
```

```python
def cap_outliers(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return series.clip(lower, upper)

for col in ["read12", "math12", "motheduc", "fatheduc", "lfaminc"]:
    if col in df.columns:
        df[col] = cap_outliers(df[col])

df.reset_index(drop=True, inplace=True)
df.head()
```

```python
df["avg_score"] = df[["read12", "math12"]].mean(axis=1)

# Convert binary indicator columns into readable labels for plotting
df["gender"] = df["female"].map({0: "Male", 1: "Female"})
df["catholic_hs"] = df["cathhs"].map({0: "No", 1: "Yes"})
df["hs_grad"] = df["hsgrad"].map({0: "No", 1: "Yes"})

def get_ethnicity(row):
    if row.get("asian", 0) == 1:
        return "Asian"
    elif row.get("hispan", 0) == 1:
        return "Hispanic"
    elif row.get("black", 0) == 1:
        return "Black"
    else:
        return "Other/White"

df["ethnicity"] = df.apply(get_ethnicity, axis=1)

# Min-Max scale family income as an example transformation
scaler = MinMaxScaler()
df["lfaminc_scaled"] = scaler.fit_transform(df[["lfaminc"]])

df[["avg_score", "gender", "catholic_hs", "hs_grad", "ethnicity", "lfaminc_scaled"]].head()
```
